```python
import torch
from transformers import CLIPProcessor, CLIPModel
from PIL import Image
import requests
import io
import matplotlib.pyplot as plt
from sklearn.metrics.pairwise import cosine_similarity
import numpy as np
import os
from glob import glob
import base64 # Required for HTML display
from IPython.display import display, HTML # Required for HTML display
import warnings
import pickle # NEW: For saving/loading the database
import hashlib # NEW: For creating a unique hash key

# 🚨 Define the path for the cached database file
DB_CACHE_PATH = '/content/gdrive/MyDrive/fashion_clip_cache.pkl'

warnings.filterwarnings('ignore') # Suppress warnings

# Define the categories for Zero-Shot Classification
CATEGORIES = [
    "A photo of a shirt",
    "A photo of a coat",
    "A photo of a jacket",
    "A photo of a skirt",
    "A photo of a pair of shoes",
    "A photo of a dress",
    "A photo of a pair of jeans",
    "A photo of a pair of socks",
    "A photo of a pair of trousers"
]

class FashionClipSearch:
    def __init__(self):
        print("Loading Fashion-CLIP model (this may take a moment)...")
        model_id = "patrickjohncyh/fashion-clip"
        self.model = CLIPModel.from_pretrained(model_id)
        self.processor = CLIPProcessor.from_pretrained(model_id)
        self.model.eval()
        self.database = []
    def get_cache_key(self, file_path):
        """Creates a simple, consistent hash from the file path."""
        return hashlib.sha256(file_path.encode('utf-8')).hexdigest()

    # 🔑 FIXED: load_image_from_url with increased redirect limit
    def load_image_from_url(self, url):
        try:
            headers = {
                'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/91.0.4472.124 Safari/537.36',
                'Accept': 'image/webp,image/apng,image/*,*/*;q=0.8'
            }
            with requests.Session() as s:
                s.max_redirects = 50
                response = s.get(url, headers=headers, stream=True, timeout=10)
                response.raise_for_status()
            return Image.open(io.BytesIO(response.content)).convert('RGB')
        except Exception as e:
            print(f"Failed to load {url}: {e}")
            return None

    # Core function: Converts image to a normalized 512D embedding vector
    def get_embedding(self, image):
        inputs = self.processor(images=image, return_tensors="pt")
        with torch.no_grad():
            image_features = self.model.get_image_features(**inputs)
        image_features = image_features / image_features.norm(p=2, dim=-1, keepdim=True)
        return image_features.numpy()

    # 🔑 NEW: Zero-Shot Classification Function
    def classify_image(self, image, category_labels):
        image_emb = self.get_embedding(image)

        inputs = self.processor(text=category_labels, return_tensors="pt", padding=True)
        with torch.no_grad():
            text_features = self.model.get_text_features(**inputs)

        text_features = text_features / text_features.norm(p=2, dim=-1, keepdim=True)
        similarities = cosine_similarity(image_emb, text_features.numpy())[0]

        best_match_index = np.argmax(similarities)
        best_category_label = category_labels[best_match_index]

        # Cleans up the label: "A photo of a skirt" -> "Skirt"
        category = best_category_label.split(" of a ")[-1].capitalize()

        return category, similarities[best_match_index]

    # 🔑 MODIFIED: Indexes files, calculates embedding, and performs classification
    def index_database_from_files(self, file_paths, categories):
        print("--- Starting Indexing with Caching ---")

        # 1. Load Existing Cache (if it exists)
        image_cache = {}
        if os.path.exists(DB_CACHE_PATH):
            try:
                with open(DB_CACHE_PATH, 'rb') as f:
                    image_cache = pickle.load(f)
                print(f"✅ Loaded {len(image_cache)} items from cache.")
            except Exception as e:
                print(f"⚠️ Could not load cache file ({e}). Starting fresh.")

        # 2. Build Database (and update cache)
        self.database = []
        newly_indexed_count = 0

        for idx, path in enumerate(file_paths):
            cache_key = self.get_cache_key(path)

            # Check the cache first!
            if cache_key in image_cache:
                item = image_cache[cache_key]
                # Restore the data from cache
                self.database.append(item)
                continue

            # If not in cache, process the image
            if idx % 100 == 0 and newly_indexed_count != 0:
                print(f"Processing image {idx}/{len(file_paths)} (New Indexing)...")

            try:
                img = Image.open(path).convert('RGB')
                emb = self.get_embedding(img)
                category, score = self.classify_image(img, categories)

                # Prepare the item data structure
                item_data = {
                    'id': idx,
                    'path': path,
                    'embedding': emb,
                    'category': category,
                    'category_score': score
                }

                # Add to current database and the cache dictionary
                self.database.append(item_data)
                image_cache[cache_key] = item_data
                newly_indexed_count += 1

            except Exception as e:
                print(f"Failed to load file {path}: {e}")

        # 3. Save Cache to Google Drive
        print(f"\nIndexed {newly_indexed_count} new items. Total items in cache: {len(image_cache)}")
        try:
            with open(DB_CACHE_PATH, 'wb') as f:
                pickle.dump(image_cache, f)
            print(f"✅ Cache successfully saved to: {DB_CACHE_PATH}")
        except Exception as e:
            print(f"🚨 Failed to save cache to Google Drive! Error: {e}")


        print(f"Successfully finalized database with {len(self.database)} items.")

    def search(self, query_url, top_k=5):
        """
        Processes a query image URL, classifies it, filters the database by category,
        and returns the top_k most similar items from the matching category.
        """
        # 1. Process Query Image and Embedding
        print(f"Processing query image from URL: {query_url}")
        query_img = self.load_image_from_url(query_url)
        if not query_img:
            return None, []

        query_emb = self.get_embedding(query_img)

        # 🔑 NEW: Classify the Query Image to determine the target category
        # Assumes CATEGORIES list is defined globally or accessible here
        global CATEGORIES
        query_category, _ = self.classify_image(query_img, CATEGORIES)
        print(f"Query classified as: **{query_category}**")

        # 2. Filter Database by Category
        db_embeddings = []
        filtered_database_items = []

        # Iterate and collect only items that match the query category
        for item in self.database:
            if item['category'] == query_category:
                filtered_database_items.append(item)
                db_embeddings.append(item['embedding'])

        if not filtered_database_items:
            print(f"Warning: No items found in the database matching category '{query_category}'. Returning empty results.")
            return query_img, []

        # Convert the filtered embeddings list to a NumPy array for comparison
        filtered_db_embeddings = np.vstack(db_embeddings)

        # 3. Calculate Cosine Similarity (The fast calculation against filtered data)
        #
        similarities = cosine_similarity(query_emb, filtered_db_embeddings).flatten()

        # 4. Sort and Select Top K
        # Get indices of the top_k scores within the *filtered* list
        top_indices = similarities.argsort()[-top_k:][::-1]

        # 5. Format Results
        results = []
        for idx in top_indices:
            # Map the index back to the original filtered item's data
            item = filtered_database_items[idx]

            results.append({
                'id': item['id'],
                'path': item['path'],
                'score': similarities[idx],
                'category': item['category'],
                'category_score': item['category_score']
            })

        return query_img, results

    # 🔑 NEW: Display results using HTML and local file paths (Most robust method)
    def display_results_by_path(self, query_img_pil, results):

        # 1. Display the Query Image
        plt.figure(figsize=(4, 4))
        plt.imshow(query_img_pil)
        plt.title("Input Query (URL)")
        plt.axis('off')
        plt.show()

        # 2. Display Matched Images using HTML and Base64
        html_content = '<div style="display: flex; flex-wrap: wrap; justify-content: start;">'

        for i, res in enumerate(results):
            # Read the binary data from the local file path
            try:
                with open(res['path'], 'rb') as f:
                    data = f.read()
                # Encode data to base64 for HTML display
                encoded_img = base64.b64encode(data).decode('utf-8')
            except FileNotFoundError:
                encoded_img = "" # Handle case where file might be deleted

            # Create the HTML structure for each image, including the score and category
            html_content += f"""
            <div style="margin: 10px; text-align: center;">
                <img src="data:image/jpeg;base64,{encoded_img}" style="width: 150px; height: 150px; object-fit: cover;">
                <p style="font-size: 10px; margin-top: 5px;">
                    Match #{i+1} | {res['category']}<br>
                    **Score: {res['score']:.3f}**<br>
                    File: {os.path.basename(res['path'])}
                </p>
            </div>
            """

        html_content += '</div>'
        display(HTML(html_content))


# --- RUNNABLE EXECUTION BLOCK ---

# 🚨 STEP 1: DEFINE YOUR PATHS
# IMPORTANT: This must match the location where you unzipped DeepFashion.
DATA_ROOT = '/content/data/deepfashion/'

# 🚨 STEP 2: FIND ALL IMAGE FILES
glob_pattern = os.path.join(DATA_ROOT, 'img', '**', '*.[jp][pn]g')
all_image_paths = glob(glob_pattern, recursive=True)

# Restrict to 1000 items for the MVP demo to save time
db_file_paths = all_image_paths[:6000]

if len(db_file_paths) == 0:
    print(f"\n🚨 ERROR: No image files found! Check the path: {glob_pattern}")
else:
    # Initialize the engine (loads the CLIP model)
    global engine
    engine = FashionClipSearch()

    # 🚨 STEP 3: INDEX AND CLASSIFY THE DATABASE
    print("--- Starting Indexing and Classification ---")
    engine.index_database_from_files(db_file_paths, CATEGORIES)

    print(f"\n✅ Indexing and Classification Complete. Database has {len(engine.database)} items.")
```
